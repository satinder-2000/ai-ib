```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
```

<b>1. Create a sample customer dataset</b>

```python
np.random.seed(42)
n= 1000

df = pd.DataFrame({
    "CustomerID": range(1001, 1001 + n),
    "Age:": np.random.randint(25, 75, n),
    "AnnualIncome": np.random.randint(30000, 250000, n),
    "AccountBalance": np.random.randint(5000, 1000000, n),
    "InvestmentAmount": np.random.randint(1000, 500000, n),
    "TransactionsPerMonth": np.random.randint(1, 50, n),
    "RiskTolerance": np.random.randint(1, 11, n),
    "InvestmentTenure": np.random.randint(1, 30, n),
    "ProductsOwned": np.random.randint(1, 10, n)
})
df.rename(columns={"Age:": "Age"}, inplace=True)
print("\ndf.columns.tolist()\n",df.columns.tolist())

#Before running the clustering model, let's conrirm that every feature is in the list

features = [
    "Age",
    "AnnualIncome",
    "AccountBalance",
    "InvestmentAmount",
    "TransactionsPerMonth",
    "RiskTolerance",
    "InvestmentTenure",
    "ProductsOwned"
]

# Check for missing feature names
missing = [col for col in features if col not in df.columns]

print("Missing features:", missing)


print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

features = [
    "Age",
    "AnnualIncome",
    "InvestmentAmount",
    "TransactionsPerMonth",
    "RiskTolerance",
    "InvestmentTenure",
    "ProductsOwned"
]

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

<b>Find optimal number of clusters using Elbow Method</b>

```python
inertia = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
    
```
